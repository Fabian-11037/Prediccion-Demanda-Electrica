# 10.16.1. Antecedentes y originalidad de M4ₙ

M4ₙ se estudia como una integración de memoria residual fuera de muestra, combinación congelada de expertos y actualización causal para pronosticar las 24 horas de demanda del SIN. La recuperación de residuos y la actualización de memoria tienen antecedentes directos; la matriz contrasta la integración temporal con esos trabajos y delimita su alcance.

## Mecanismo que se contrasta

Los componentes de M4ₙ que se comparan con la literatura son los de la ficha de la arquitectura:

| Código | Componente |
|---|---|
| (a) | Base congelada: combinación aprendida de PatchTST, el experto E y el experto periódico (M1) |
| (b) | Memoria de residuos normalizados de predicciones fuera de muestra, generadas con OOF temporal anidado por prefijos |
| (c) | Recuperación con proyecciones aprendidas de consulta y clave sobre un contexto de 912 componentes, 16 vecinos y máscara de disponibilidad de 2 a 365 días |
| (d) | Compuerta escalar que regula la corrección según semejanza, dispersión, respaldo y antigüedad de los vecinos |
| (e) | Banco que se actualiza durante la evaluación con residuos ya revelados, con los pesos congelados |
| (f) | Pérdida absoluta normalizada por q, comparable con MASE; demanda del SIN con 24 horas directas |

Las versiones exactas para arXiv y Semantic Scholar, con su sintaxis, están en el script de búsqueda.

````python
from pathlib import Path
import json
import pandas as pd
from IPython.display import display
D = next(c for p in [Path.cwd(), *Path.cwd().parents] for c in (p/'_static/literature/novedad', p/'book/_static/literature/novedad') if c.exists())
bus = json.loads((D/'busqueda_novedad_20261125.json').read_text(encoding='utf-8'))
actualizacion = json.loads((D/'paginacion_20261126.json').read_text(encoding='utf-8'))
sno = json.loads((D/'snowball_repos_20261125.json').read_text(encoding='utf-8'))
c9 = pd.read_csv(D/'c9_corpus_scopus_wos_ieee.csv', dtype=str)
ampliadas = {(q['base'],q['cadena']):q for q in actualizacion['consultas']}
filas=[]
for q in bus['consultas']:
    nuevo=ampliadas.get((q['base'],q['cadena']))
    filas.append({'Base':q['base'],'Cadena':q['cadena'],'Total índice · 25nov':q['resultados'],'Recuperados · 25nov':len(q['registros']),'Total índice · 26nov':nuevo.get('total_indice') if nuevo else None,'Recuperados · 26nov':nuevo['recuperados'] if nuevo else None,'Paginación adicional':('agotada' if nuevo['paginacion_agotada'] else 'parcial') if nuevo else 'sin ampliación'})
inventario = pd.DataFrame(filas)
display(inventario)
assert len(ampliadas)==5
````

La tabla distingue el total informado por cada índice de los registros efectivamente recuperados. La captura inicial tenía un máximo de 50 por consulta. La ampliación agota la paginación de cinco consultas el 26 de noviembre; los totales pueden cambiar entre fechas por actualización del índice. La recuperación completa no equivale a revisión independiente de todos los textos.

La ampliación incorpora *FreshCast*, ya revisado en texto completo en el estudio, y localiza *Decouple to Reuse*, con bancos residuales de representaciones. Este último se clasifica con alcance de resumen y no se equipara a una memoria de errores de pronóstico.

## Contribución delimitada y contraste experimental

RATL establece el antecedente directo de recuperación de residuos de una base congelada; FreshCast establece el de memoria que incorpora observaciones reveladas. La contribución candidata de M4 se sitúa en la construcción OOF temporal anidada de los residuos de una combinación de expertos y su recuperación y actualización con disponibilidad definida para el origen diario del SIN.

Las {ref}`ablaciones de M4 <complementos-m4>` respaldan la utilidad de actualizar el banco frente a mantenerlo fijo. No acreditan beneficio de la compuerta y la recuperación por distancia numérica obtiene menor MASE que la aprendida. Por ello, ni la compuerta ni el aprendizaje de semejanza se presentan como mejoras necesarias. La construcción OOF es una propiedad del esquema; este contraste no aísla por sí solo su efecto.

Los resultados sustentan conclusiones de desempeño dentro del benchmark. La originalidad requiere además contrastar mecanismo y entrenamiento con los antecedentes, y no se deriva únicamente del territorio de aplicación ni de superar a otros modelos.

## Declaración y alcance de originalidad

En los trabajos examinados no se identificó la misma integración de residuos OOF temporales de una combinación congelada de expertos, recuperación contextual y actualización causal para demanda nacional directa de24 horas. Esta conclusión se restringe a la evidencia y al alcance de las búsquedas documentadas; no afirma prioridad mundial del principio de recuperar residuos ni de actualizar memoria.

La consulta C9 examina el inventario del anexo PRISMA, no búsquedas nativas del mecanismo en Scopus, Web of Science e IEEE. Las clasificaciones basadas únicamente en resumen se mantienen identificadas. La revisión no aporta un segundo evaluador humano independiente; la ampliación de registros no sustituye esa verificación.
