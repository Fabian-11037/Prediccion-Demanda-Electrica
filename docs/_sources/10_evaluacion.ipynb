MLP presenta el menor MASE medio del piloto externo (0,542); N-HiTS registra 0,827. PatchTST utiliza la precisión de parada interna, presentada por separado. La dispersión entre semillas no sigue necesariamente ese orden; un error medio mayor puede ser estable entre inicializaciones.

Para expresar las diferencias en una escala relativa común se conserva como referencia la media piloto de MLP, obtenida exclusivamente de validación. Con $d_s=(M_{A,s}-M_{B,s})/\overline M_{\mathrm{MLP,piloto}}$, margen $\delta=0,03$, $\alpha=0,05$ y potencia objetivo $0,80$, se calcula

$$R\approx\left\lceil\left[\frac{(z_{1-\alpha/2}+z_{0,80})\widehat\sigma_d}{\delta}\right]^2\right\rceil,\qquad
\mathrm{MDD}_5=(z_{1-\alpha/2}+z_{0,80})\frac{\widehat\sigma_d}{\sqrt5}.$$

Se informa además $\delta/\widehat\sigma_d$ y una planificación conservadora con Bonferroni para las diez parejas. El mínimo de diseño es cinco semillas. Es una aproximación retrospectiva de variación entre inicializaciones, condicionada a los mismos cuatro años; no estima la incertidumbre de nuevas fechas ni acredita la potencia alcanzada por los ajustes ya ejecutados.

````python
from itertools import combinations
g_delta=.03;g_alpha=.05;g_power=.80
g_pilot_reference=float(g_pilot_seed['mlp'].mean())
g_pilot_pairs=list(combinations(g_pilot_seed.columns,2))
assert len(g_pilot_pairs)==10
g_z=stats.norm.ppf(1-g_alpha/2)+stats.norm.ppf(g_power)
g_z_bonf=stats.norm.ppf(1-g_alpha/(2*len(g_pilot_pairs)))+stats.norm.ppf(g_power)
g_dimension=[]
for g_a,g_b in g_pilot_pairs:
    g_difference=(g_pilot_seed[g_a]-g_pilot_seed[g_b])/g_pilot_reference
    g_sd=float(g_difference.std(ddof=1))
    g_dimension.append({'A':g_a,'B':g_b,'Comparación':g_names[g_a]+' − '+g_names[g_b],
        'DE diferencia relativa':g_sd,'δ / DE':g_delta/g_sd if g_sd>0 else np.inf,
        'R individual':max(5,int(np.ceil((g_z*g_sd/g_delta)**2))),
        'R Bonferroni(10)':max(5,int(np.ceil((g_z_bonf*g_sd/g_delta)**2))),
        'MDD con 5 (% referencia piloto)':100*g_z*g_sd/np.sqrt(5)})
g_dimension=pd.DataFrame(g_dimension)
display(g_dimension.drop(columns=['A','B']).round(5))
g_dimension.to_csv(g_result/'dimensionamiento_piloto_integrado.csv',index=False)
````

La aproximación individual exige entre cinco y 45 semillas por procedimiento, según la pareja; el máximo conservador para las diez comparaciones es 76. Las cinco semillas ejecutadas no acreditan potencia de 0,80 para el margen del 3 % en todas las parejas. Los ocho procedimientos sin piloto independiente conservan su dimensionamiento retrospectivo de parada interna, presentado por separado.

N-HiTS conserva además un diagnóstico temporal por bootstrap de bloques de 14 días frente al ingenuo semanal de validación. Su estimación límite es 0,0843 y no determina un número final de semillas. Ese diagnóstico conserva incertidumbre temporal y no equivale al cálculo anterior entre inicializaciones.

### Precisión en la reserva interna

TFT, TCN, iTransformer, el control numérico N, TaTS con texto oficial, fusión tardía y las variantes N y N+O de TGForecaster conservan métricas de parada de sus cinco entrenamientos finales, semillas 1000–1004. Se utilizan para completar el análisis de precisión de desarrollo. Los ocho procedimientos comparten 803 orígenes de reserva interna, del 21 de octubre de 2019 al 31 de diciembre de 2021, y el denominador MASE q168 de 215.617,90361589016 kWh.

Se calcula media, DE, CV e IC95 % t entre semillas sobre el MASE de parada del checkpoint restaurado. La reserva seleccionó también la época de parada; los resultados son descriptivos y retrospectivos. Los intervalos condicionan ese periodo y suponen independencia y normalidad aproximada entre inicializaciones, difíciles de evaluar con cinco realizaciones.

````python
# Métricas de desarrollo guardadas; no se utilizan errores de prueba.
from itertools import combinations
p_inputs=pd.read_csv(g_result/'entradas_precision_parada.csv')
assert len(p_inputs)==5*p_inputs.clave.nunique() and p_inputs.n_parada.eq(803).all()
def calcular_precision_parada(entradas,referencia='adj_modelo6_20261112_1035__N'):
    """Usa solo MASE de desarrollo; no estima potencia con el efecto observado."""
    required=['clave','semilla','MASE_parada','n_parada','inicio_parada','fin_parada','q168_kwh']
    x=entradas[required].copy()
    assert not x.isna().any().any()
    assert not x.duplicated(['clave','semilla']).any()
    assert np.isfinite(x.MASE_parada).all() and x.MASE_parada.ge(0).all()
    assert x[['n_parada','inicio_parada','fin_parada','q168_kwh']].drop_duplicates().shape[0]==1
    for _,g in x.groupby('clave'):
        assert sorted(g.semilla.tolist())==list(range(1000,1005))
    matriz=x.pivot(index='semilla',columns='clave',values='MASE_parada').sort_index()
    assert matriz.notna().all().all() and referencia in matriz
    n=len(matriz);alpha=.05;objetivo=.80;delta=.03
    tcrit=stats.t.ppf(1-alpha/2,n-1)
    resumen=matriz.agg(['mean','std','count']).T.rename(columns={'mean':'MASE_media','std':'DE','count':'semillas'})
    resumen['CV_pct']=100*resumen.DE/resumen.MASE_media
    resumen['IC95_inf']=resumen.MASE_media-tcrit*resumen.DE/np.sqrt(n)
    resumen['IC95_sup']=resumen.MASE_media+tcrit*resumen.DE/np.sqrt(n)
    mu_ref=float(resumen.loc[referencia,'MASE_media'])
    assert mu_ref>0
    parejas=list(combinations(matriz.columns,2))
    z=stats.norm.ppf(1-alpha/2)+stats.norm.ppf(objetivo)
    z_bonf=stats.norm.ppf(1-alpha/(2*len(parejas)))+stats.norm.ppf(objetivo)
    filas=[]
    for a,b in parejas:
        d=matriz[a]-matriz[b]
        media=float(d.mean());sd=float(d.std(ddof=1));sd_rel=sd/mu_ref
        margen=tcrit*sd/np.sqrt(n)
        filas.append(dict(A=a,B=b,semillas=n,diferencia_MASE=media,DE_diferencia=sd,
            IC95_diferencia_inf=media-margen,IC95_diferencia_sup=media+margen,
            diferencia_pct_referencia=100*media/mu_ref,DE_relativa=sd_rel,
            referencia=referencia,MASE_referencia=mu_ref,delta_relativo=delta,
            alpha=alpha,potencia_objetivo=objetivo,parejas_familia=len(parejas),
            delta_sobre_DE=delta/sd_rel if sd_rel>0 else np.inf,
            R_individual=max(n,int(np.ceil((z*sd_rel/delta)**2))),
            R_Bonferroni=max(n,int(np.ceil((z_bonf*sd_rel/delta)**2))),
            MDD_5_pct_referencia=100*z*sd_rel/np.sqrt(n)))
    return resumen.reset_index(),pd.DataFrame(filas),matriz

p_summary,p_pairs,p_matrix=calcular_precision_parada(p_inputs)
p_summary.insert(1,'Procedimiento',p_summary.clave.map(g_names))
p_pairs.insert(2,'Comparación',p_pairs.A.map(g_names)+' − '+p_pairs.B.map(g_names))
assert np.allclose(p_summary.set_index('clave').MASE_media,
    v_precision.MASE_media.reindex(p_summary.clave))
````

TaTS adaptado · N+O registra el menor MASE de parada medio (0.565). El CV oscila entre 2.74 % en iTransformer · N y 18.61 % en TGForecaster · N+O. La estabilidad entre semillas y el error medio describen propiedades diferentes; el resumen integrado no sustituye el benchmark de prueba.

La planificación conserva δ=3 %, α=0,05 y potencia objetivo de 0,80. Las diferencias pareadas se expresan respecto de la media de parada del control N (0.570981), común a los ocho procedimientos. Se utiliza la misma aproximación normal y MDD con cinco semillas descritas antes, con Bonferroni para las veintiocho parejas. El cálculo usa la DE de las diferencias y el margen fijado, sin estimar potencia a partir de la diferencia observada.

````python
p_columns=['Comparación','diferencia_MASE','DE_diferencia','IC95_diferencia_inf','IC95_diferencia_sup',
    'DE_relativa','delta_sobre_DE','R_individual','R_Bonferroni','MDD_5_pct_referencia']
display(p_pairs[p_columns].rename(columns={'diferencia_MASE':'Diferencia MASE','DE_diferencia':'DE diferencia',
    'IC95_diferencia_inf':'IC95 inferior','IC95_diferencia_sup':'IC95 superior',
    'DE_relativa':'DE relativa','delta_sobre_DE':'δ / DE','R_individual':'R individual',
    'R_Bonferroni':f'R Bonferroni({len(p_pairs)})','MDD_5_pct_referencia':'MDD con 5 (% control N)'}).round(5))
assert len(p_pairs)==len(p_summary)*(len(p_summary)-1)//2 and set(p_pairs.A)|set(p_pairs.B)==set(p_summary.clave)
````

El máximo requerido por la aproximación individual es 2112 semillas y asciende a 4231 con Bonferroni para las veintiocho parejas. Las cinco ejecuciones disponibles permiten estimar variabilidad, pero no acreditan potencia de 0,80 para el margen del 3 % en todas las comparaciones. Las estimaciones son sensibles a cinco diferencias de desarrollo y no garantizan la potencia de una evaluación externa.

````python
v_dimension_groups=[(g_dimension,'A','B','R Bonferroni(10)','Piloto externo · referencia MLP'),
    (p_pairs,'A','B','R_Bonferroni','Parada interna · referencia control N')]
from matplotlib.colors import LogNorm
fig,axes=plt.subplots(1,2,figsize=(18,8),layout='constrained')
v_max=max(float(g_dimension['R Bonferroni(10)'].max()),float(p_pairs.R_Bonferroni.max()))
for ax,(v_df,v_a,v_b,v_r,v_title) in zip(axes,v_dimension_groups):
    v_ks=[k for k in g_keys if k in set(v_df[v_a])|set(v_df[v_b])]
    v_matrix=np.full((len(v_ks),len(v_ks)),np.nan)
    for _,v_row in v_df.iterrows():
        v_i,v_j=v_ks.index(v_row[v_a]),v_ks.index(v_row[v_b])
        v_matrix[v_i,v_j]=v_matrix[v_j,v_i]=v_row[v_r]
    v_im=ax.imshow(v_matrix,cmap='YlOrRd',norm=LogNorm(vmin=5,vmax=v_max))
    ax.set(title=v_title,xticks=range(len(v_ks)),yticks=range(len(v_ks)),
        xticklabels=[g_names[k] for k in v_ks],yticklabels=[g_names[k] for k in v_ks])
    ax.tick_params(axis='x',rotation=60,labelsize=8);ax.tick_params(axis='y',labelsize=8);ax.grid(False)
    for v_i in range(len(v_ks)):
        for v_j in range(len(v_ks)):
            if v_i!=v_j:ax.text(v_j,v_i,f'{v_matrix[v_i,v_j]:.0f}',ha='center',va='center',fontsize=10,
                color='white' if v_matrix[v_i,v_j]>.6*v_max else '#222222')
fig.colorbar(v_im,ax=axes.tolist(),shrink=.65,label='Semillas aproximadas · escala logarítmica; Bonferroni por grupo')
fig.savefig(g_figures/'dimensionamiento_fuentes.png',dpi=160)
display(fig);plt.close(fig)
````

$$I_g=\frac{1}{50}\sum_{s=1}^{5}\sum_{r=1}^{10}

Las figuras muestran el cambio porcentual respecto del MAE basal de cada modelo, equivalente al cambio relativo de MASE con el denominador fijo. ### Alcance de la atribución

Las tres codificaciones de hora son idénticas entre orígenes diarios iniciados a medianoche. Permutar orígenes deja ese bloque inalterado; su importancia no es identificable mediante este procedimiento y no se reporta como cero. La salida directa de 24 posiciones también representa la posición horaria.

El desplazamiento preserva las trayectorias de cada bloque, pero rompe dependencias entre demanda, calendario y objetivo. Por ello mide sensibilidad ante perturbaciones fuera de la combinación observada y no estima el efecto de retirar una variable y reentrenar. Los grupos correlacionados pueden sustituirse parcialmente; una importancia pequeña no acredita irrelevancia intrínseca. La prueba conserva su alcance retrospectivo y estos resultados no modifican arquitecturas, selección de entradas ni métricas del benchmark.

### Pipeline reproducible de inferencia y perturbación

El siguiente código utiliza los checkpoints y escaladores conservados, verifica sus huellas, reproduce el baseline y calcula las perturbaciones. La construcción documental lee sus CSV; el cálculo se ejecuta explícitamente con `python scripts/importancia_variables.py` en un entorno con PyTorch CPU. No contiene pasos de optimización.

```python
"""Importancia agrupada mediante inferencia de checkpoints congelados."""
from pathlib import Path
from types import SimpleNamespace
import argparse
import hashlib
import json
import platform
import sys
import time
import numpy as np
import pandas as pd

ROOT = Path(__file__).resolve().parents[1]
MODELOS = ['mlp', 'arx_lineal', 'cnn1d', 'lstm']
REPETICIONES = 10
SEMILLA_PERMUTACION = 11037
GRUPOS_CAL = {
    'semanal_festivos': ['cal_dow', 'cal_es_festivo', 'cal_es_domingo',
        'cal_es_sabado', 'cal_es_no_laboral', 'cal_tipo_dia',
        'cal_dias_al_siguiente_festivo', 'cal_dias_desde_festivo_previo',
        'cal_es_vispera_festivo', 'cal_es_dia_post_festivo',
        'cal_es_fin_semana_largo', 'cal_dow_sin', 'cal_dow_cos'],
    'anual_temporadas': ['cal_mes', 'cal_dia_anio', 'cal_es_semana_santa',
        'cal_es_temporada_navidena', 'cal_es_fin_de_anio', 'cal_doy_sin', 'cal_doy_cos'],
    'hora': ['cal_hora', 'cal_hora_sin', 'cal_hora_cos'],
}
GRUPOS = ['demanda_completa', 'perfil_demanda', 'nivel_ventana',
          'calendario_completo', 'semanal_festivos', 'anual_temporadas'] + [f'dia_{i}' for i in range(1, 8)]


def sha256(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()


def indices_donantes(fechas, repeticion):
    """Traslada circularmente orígenes completos dentro de cada año."""
    rng = np.random.default_rng(SEMILLA_PERMUTACION + repeticion)
    perm = np.arange(len(fechas))
    for anio in sorted(set(fechas.year)):
        ix = np.flatnonzero(fechas.year == anio)
        if len(ix) < 57:
            raise ValueError('Se requieren al menos 57 orígenes diarios por año')
        desplazamiento = int(rng.integers(28, len(ix) - 27))
        perm[ix] = np.roll(ix, desplazamiento)
    assert np.array_equal(np.sort(perm), np.arange(len(fechas)))
    assert np.array_equal(fechas.year[perm], fechas.year)
    assert np.all(perm != np.arange(len(fechas)))
    return perm


def perturbar(xp, cf, nivel, donantes, grupo, columnas_cal):
    """Copia solo el grupo perturbado; las etiquetas nunca entran aquí."""
    if grupo not in GRUPOS:
        raise ValueError(f'Grupo no reconocido: {grupo}')
    x, c, n = xp, cf, nivel
    if grupo in ('demanda_completa', 'perfil_demanda'):
        x = xp.copy()
        x[:, :, 0] = xp[donantes, :, 0]
    if grupo in ('demanda_completa', 'nivel_ventana'):
        n = nivel[donantes]
    if grupo.startswith('dia_'):
        dia = int(grupo.split('_')[1])
        lo, hi = xp.shape[1] - 24 * dia, xp.shape[1] - 24 * (dia - 1)
        x = xp.copy()
        x[:, lo:hi, 0] = xp[donantes, lo:hi, 0]
    if grupo in ('calendario_completo', 'semanal_festivos', 'anual_temporadas'):
        cols = columnas_cal if grupo == 'calendario_completo' else GRUPOS_CAL[grupo]
        ix = [columnas_cal.index(k) for k in cols]
        x = xp.copy()
        c = cf.copy()
        # xp concatena demanda y calendario pasado; cf conserva calendario futuro.
        x[:, :, np.asarray(ix) + 1] = xp[donantes][:, :, np.asarray(ix) + 1]
        c[:, :, ix] = cf[donantes][:, :, ix]
    return x, c, n


def restaurar_entradas(panel, fechas, escalador):
    """Usa solo escaladores publicados; no ajusta estadísticas."""
    cols = escalador['columnas_cal']
    assert escalador['columnas_x'] == ['x_demanda_real_kwh']
    assert escalador['datos']['ventana'] == 168
    assert escalador['datos']['normalizacion'] == 'nivel'
    assert pd.Timestamp(escalador['fin_ajuste']) < pd.Timestamp('2022-01-01', tz='America/Bogota')
    pos = panel.index.get_indexer(fechas)
    assert (pos >= 168).all()
    pasada = pos[:, None] - np.arange(168, 0, -1)
    futura = pos[:, None] + np.arange(24)
    assert np.all(panel.index.asi8[pasada[:, -1]] < fechas.asi8)
    historia = panel['x_demanda_real_kwh'].to_numpy(dtype=np.float64)[pasada]
    nivel = historia.mean(axis=1)
    x = (historia / nivel[:, None]).astype('float32')
    if escalador['secuencial']:
        x = (x - 1.) / escalador['sigma_demanda']
    media = np.array([escalador['media'][k] for k in cols])
    std = np.array([escalador['std'][k] for k in cols])
    cal = panel[cols].to_numpy(dtype=np.float64)
    cp = ((cal[pasada] - media) / std).astype('float32')
    cf = ((cal[futura] - media) / std).astype('float32')
    y = panel['y_demanda_real_kwh'].to_numpy()[futura]
    xp = np.concatenate([x[:, :, None], cp], axis=2)
    assert xp.dtype == cf.dtype == np.float32
    assert np.isfinite(xp).all() and np.isfinite(cf).all() and (nivel > 0).all()
    indices_hora = [cols.index(k) for k in GRUPOS_CAL['hora']]
    assert all(np.array_equal(cp[:, :, j], np.broadcast_to(cp[0:1, :, j], cp[:, :, j].shape)) for j in indices_hora)
    assert all(np.array_equal(cf[:, :, j], np.broadcast_to(cf[0:1, :, j], cf[:, :, j].shape)) for j in indices_hora)
    return xp, cf, nivel, y, cols


def predecir(red, xp, cf, nivel, torch):
    salida = []
    for i in range(0, len(nivel), 256):
        yhat = red(torch.from_numpy(xp[i:i+256]), torch.from_numpy(cf[i:i+256]))
        salida.append(yhat.cpu().numpy())
    pred = np.concatenate(salida).astype(np.float64) * nivel[:, None]
    assert pred.shape == (len(nivel), 24) and np.isfinite(pred).all()
    return pred


def ejecutar(verificar=False):
    import torch
    torch.set_num_threads(1)
    torch.set_num_interop_threads(1)
    torch.use_deterministic_algorithms(True)
    torch.manual_seed(SEMILLA_PERMUTACION)
    sha = json.loads((ROOT/'results/jbook/modelos_actualizados.json').read_text())['commit']
    snapshot = ROOT/'results/repositorio'/sha
    sys.path.insert(0, str(snapshot/'src'))
    from demanda import modelos_base as M
    from demanda.reentrenamiento_v3 import Secuencial as CNN
    from demanda.variantes_v4 import Secuencial as LSTM
    registro = json.loads((snapshot/'results/consolidado/modelos.json').read_text(encoding='utf-8'))
    huellas = {r['archivo']:r['sha256'] for r in json.loads((snapshot/'IMPORTACION.json').read_text(encoding='utf-8'))['archivos']}
    out = ROOT/'results/jbook/importancia_variables'
    out.mkdir(parents=True, exist_ok=True)
    protocolo = dict(version=1, escenario='diagnostico_retrospectivo_prueba_2022_2025',
        modelos=MODELOS, semillas=list(range(1000,1005)), repeticiones=REPETICIONES,
        semilla_permutacion=SEMILLA_PERMUTACION, grupos=GRUPOS,
        grupos_calendario=GRUPOS_CAL, donantes='desplazamiento_circular_por_anio',
        desplazamiento_minimo_dias=28, denominador_mase=215617.90361589016,
        dispositivo='cpu', hilos=1, batch=256, entrenamiento=False,
        reajuste_escaladores=False, seleccion_con_resultados=False,
        hora='invariante_entre_origenes_no_identificable_con_esta_permutacion')
    p = out/'protocolo.json'
    if p.exists():
        assert json.loads(p.read_text(encoding='utf-8'))['protocolo'] == protocolo
    else:
        p.write_text(json.dumps(dict(registrado=pd.Timestamp.now(tz='America/Bogota').isoformat(),
            protocolo=protocolo), ensure_ascii=False, indent=2), encoding='utf-8')
    primer = registro['mlp']['artefactos'][0]
    with np.load(snapshot/primer['predicciones'], allow_pickle=False) as z:
        fechas = pd.to_datetime(z['origen'], unit='ns', utc=True).tz_convert('America/Bogota')
    assert fechas.equals(pd.date_range('2022-01-01','2025-12-31',freq='D',tz='America/Bogota'))
    cols = json.loads((snapshot/primer['escaladores']).read_text(encoding='utf-8'))['columnas_cal']
    assert sorted(sum(GRUPOS_CAL.values(), [])) == sorted(cols)
    panel = pd.read_parquet(ROOT/'data/processed/panel_horario.parquet', columns=['x_demanda_real_kwh','y_demanda_real_kwh',*cols])
    donantes = [indices_donantes(fechas,r) for r in range(REPETICIONES)]
    np.savez_compressed(out/'donantes.npz', fechas_utc=fechas.asi8, indices=np.stack(donantes))
    benchmark_path = ROOT/'results/jbook/modelos_actuales/benchmark.csv'
    benchmark_hash = sha256(benchmark_path)
    filas, validaciones = [], []
    inicio = time.perf_counter()
    with torch.inference_mode():
        for nombre in MODELOS:
            for a in sorted(registro[nombre]['artefactos'], key=lambda x:x['semilla']):
                for campo in ('pesos','escaladores','predicciones'):
                    assert sha256(snapshot/a[campo]) == huellas[a[campo]]
                esc = json.loads((snapshot/a['escaladores']).read_text(encoding='utf-8'))
                xp, cf, nivel, y, cols = restaurar_entradas(panel, fechas, esc)
                shapes = SimpleNamespace(x=xp[:, :, :1], cal_p=xp[:, :, 1:], cal_f=cf)
                params = registro[nombre]['configuracion']['params']
                red = (LSTM(nombre, params, shapes) if nombre=='lstm' else
                       CNN(nombre, params, shapes) if nombre=='cnn1d' else
                       M.construir_red(nombre, params, shapes))
                ck = torch.load(snapshot/a['pesos'], map_location='cpu', weights_only=True)
                red.load_state_dict(ck['state_dict'], strict=True)
                red.eval()
                esperado = np.load(snapshot/a['predicciones'], allow_pickle=False)
                assert np.array_equal(esperado['y'], y)
                assert np.array_equal(esperado['origen'], fechas.asi8)
                base = predecir(red, xp, cf, nivel, torch)
                error = np.abs(base-y)
                mae = error.mean()
                diferencia = float(np.max(np.abs(base-esperado['yhat'])))
                diferencia_mae = float(abs(mae-np.abs(esperado['yhat']-y).mean()))
                assert diferencia < 100 and diferencia_mae < 5, (nombre, diferencia, diferencia_mae)
                validaciones.append(dict(modelo=nombre, semilla=a['semilla'],
                    max_diferencia_pred_kWh=diferencia, diferencia_MAE_kWh=diferencia_mae,
                    MAE_cpu_kWh=float(mae), checkpoint_sha256=huellas[a['pesos']]))
                if not verificar:
                    for grupo in GRUPOS:
                        if nombre=='arx_lineal' and grupo.startswith('dia_') and int(grupo[4:])>2:
                            for r in range(REPETICIONES):
                                filas.append(dict(modelo=nombre,semilla=a['semilla'],grupo=grupo,
                                    repeticion=r,usa_grupo=False,MAE_base=mae,MAE_perturbado=mae,
                                    delta_MAE=0.,delta_MASE=0.,delta_rel_pct=0.))
                            continue
                        for r, perm in enumerate(donantes):
                            x, c, n = perturbar(xp, cf, nivel, perm, grupo, cols)
                            pred = predecir(red, x, c, n, torch)
                            nuevo = float(np.abs(pred-y).mean())
                            filas.append(dict(modelo=nombre,semilla=a['semilla'],grupo=grupo,
                                repeticion=r,usa_grupo=True,MAE_base=mae,MAE_perturbado=nuevo,
                                delta_MAE=nuevo-mae,delta_MASE=(nuevo-mae)/esc['escala_mase'],
                                delta_rel_pct=100*(nuevo/mae-1)))
                pd.DataFrame(validaciones).to_csv(out/'reproduccion_baseline.csv', index=False)
                if filas:
                    pd.DataFrame(filas).to_csv(out/'permutaciones.csv', index=False)
                print(f"{nombre} semilla {a['semilla']}: {'verificado' if verificar else 'completo'}; {time.perf_counter()-inicio:.1f} s", flush=True)
    assert sha256(benchmark_path) == benchmark_hash
    reporte = dict(completo=True,solo_verificacion=verificar,horas=35064,origenes=1461,
        modelos=4,checkpoints=20,repeticiones=REPETICIONES,grupos=len(GRUPOS),
        filas=len(filas),entrenamiento=False,publicado=False,benchmark_intacto=True,
        max_diferencia_pred_kWh=max(r['max_diferencia_pred_kWh'] for r in validaciones),
        max_diferencia_MAE_kWh=max(r['diferencia_MAE_kWh'] for r in validaciones),
        duracion_segundos=time.perf_counter()-inicio,torch=torch.__version__,
        numpy=np.__version__,pandas=pd.__version__,sistema=platform.platform(),
        codigo_sha256=sha256(Path(__file__)))
    (out/('validacion.json' if verificar else 'ejecucion.json')).write_text(json.dumps(reporte, ensure_ascii=False, indent=2), encoding='utf-8')
    print(json.dumps(reporte,ensure_ascii=False),flush=True)


if __name__=='__main__':
    parser=argparse.ArgumentParser()
    parser.add_argument('--verificar', action='store_true')
    ejecutar(parser.parse_args().verificar)
```

## Discusión y alcance del Entregable 2

El mejor resultado disponible pertenece al conjunto TaTS–control numérico, no a un ganador estadístico único. Añadir texto oficial no acredita una mejora práctica del 3 % en TaTS y empeora la fusión tardía y TGForecaster. La equivalencia temporal de TaTS con su control sugiere valorar el coste de codificación y parámetros adicionales antes de recomendar el componente textual. No demuestra que todo texto sea inútil: el resultado está limitado al corpus declarado, sus fechas, selección y mecanismos de fusión.

La comparación de coste debe utilizar parámetros, tiempos y dispositivos registrados; GPU AMD/ROCm y NVIDIA/CUDA no son condiciones idénticas. No se comparan tiempos de entrenamiento como si midieran exclusivamente eficiencia arquitectónica. Las curvas y diagnósticos muestran error remanente; no permiten estimar un ruido irreducible ni afirmar cercanía a un techo de desempeño.

Las pruebas incorporadas cubren la evaluación estadística posible con las predicciones recibidas. El E2 no queda completamente satisfecho: faltan una referencia ARIMA/ETS, presupuestos homogéneos, dimensionamiento suficiente de semillas, las búsquedas arquitectónicas exigibles de §14 y ablaciones/robustez controladas que no se reconstruyen con predicciones finales. Clark–West no se aplica automáticamente a MAE ni a estas redes anidadas no lineales; Giacomini–White requiere justificar su esquema de estimación e instrumentos. Estas omisiones no se reemplazan por p-valores de otra hipótesis.

La selección y exposición retrospectiva limitan la validez confirmatoria. Los IC temporales son condicionales a las cinco semillas; Friedman y Wilcoxon caracterizan incertidumbre de inicialización en esta serie, no generalización entre datasets. Las amenazas de disponibilidad textual, cambios de composición y revisiones de reanálisis se conservan en el diseño de datos.

La precisión de desarrollo está estimada para los trece procedimientos: cinco mediante piloto externo y ocho mediante parada interna. Esto completa la descripción de variación y la planificación retrospectiva; no acredita por sí solo potencia de 0,80 ni convierte la reserva de selección de época en una evaluación externa.

El conjunto principal comprende diez modelos entrenados: ARX, MLP, LSTM, CNN1D, TFT, TCN, iTransformer, TaTS, fusión tardía y TGForecaster. PatchTST y N-HiTS se conservan como comparadores numéricos; las dos variantes de TGForecaster llevan el total a trece procedimientos, además de las dos referencias deterministas. Tampoco se documenta una réplica del paper guía bajo su dataset, partición y métrica originales (§5). Los parámetros y la latencia están registrados para todos los procedimientos; solo N-HiTS conserva un registro de memoria pico y no se reportan FLOPs para los trece procedimientos (§4.2).

```python
    assert sha(CON/m['artefactos']['pesos'])==m['sha256_checkpoint']
```
